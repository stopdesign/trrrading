```python
import pandas as pd

pd.options.display.width = 120


ticker = "USD:CAD"

src = get_exante_data(ticker, start="2021-01-01 00:00+0", end="2021-03-01 00:00+0")
src = resample_ohlc(src, "1d")

print("get done")
```

```python
import talib
from talipp.indicators import EMA, SMA, Stoch, ATR

df = src.head(30).copy(deep=True)


ema = EMA(period=3, input_values=[])
atr = ATR(period=3, input_values=[])

df["ema"] = None

for dt, row in df.iterrows():
    ema.add_input_value(row.close)
    atr.add_input_value(row)

    if ema:
        df.at[dt, "ema"] = ema[-1]
    if atr:
        df.at[dt, "atr"] = atr[-1]


df["atr_orig"] = talib.ATR(df["high"], df["low"], df["close"], timeperiod=3)

atr_2 = [None] + list(ATR(period=3, input_values=list(df.itertuples())))

df["atr_2"] = atr_2
print(df)
# print(list(df.itertuples()))
```
